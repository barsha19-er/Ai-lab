## Overview of PyTorch Tensor Creation Functions

i. torch.rand(shape) – Creates a tensor with random numbers that are uniformly distributed between 0 and 1.

ii. torch.zeros(shape) – Returns a tensor of the specified shape where every value is set to 0.

iii. torch.ones(shape) – Generates a tensor of the given shape with all elements initialized to 1.

iv. torch.eye(n) – Creates an identity matrix of size n × n, with 1s along the main diagonal and 0s elsewhere.

v. torch.arange(start, end, step) – Produces a sequence of values beginning at start and increasing by step until it reaches end (excluding end).

vi. torch.linspace(start, end, steps) – Generates a tensor containing a fixed number of evenly spaced values between start and end, including both endpoints.

```python
rand = torch.rand(3,4)      
zeros = torch.zeros(3,4)    
ones = torch.ones(3,4)      
eye = torch.eye(3)          

arange = torch.arange(start=0, end=10, step=2)
linspace = torch.linspace(start=0, end=10, steps=5)

print(f"rand: {rand}")
print(f"zeros: {zeros}")
print(f"ones: {ones}")
print(f"eye: {eye}")
print(f"arange: {arange}")
print(f"linspace: {linspace}")
```

## Tensor Arithmetic in PyTorch
* Operator (+) vs torch.add()

Both the + operator and torch.add() function perform element-wise addition between tensors. However, torch.add() provides additional options that make it more flexible in certain situations.

### Operator (+)
i.Simple and Python-friendly syntax

ii.Performs element-wise addition

iii.Creates a new tensor for the result

iv.Best suited for quick and readable operations

### torch.add() Function
i.More flexible and configurable

ii.Also performs element-wise addition

iii.Supports extra features like:Output control using out,Scaling using alpha  



```python
a = torch.tensor([1, 2, 3])
b = torch.tensor([3, 4, 5])

print(a+b)
print(a-b)
print(a*b)

# adding using add()
print(torch.add(a,b))

print(torch.add(a, b, alpha=2))
```

### @ vs. * for Tensor Multiplication

When working with PyTorch tensors, the * and @ operators perform different types of multiplication.

#### * (Element-wise Multiplication)

i. The * operator multiplies tensors element by element.

ii. The tensors must have the same shape or be compatible through broadcasting.

iii. Each value in one tensor is multiplied by the corresponding value in the other tensor.

iv. This type of multiplication is often called the Hadamard product.

#### @ (Matrix Multiplication)

i. The @ operator is used for matrix multiplication.

ii. For two matrices, the number of columns in the first matrix must equal the number of rows in the second matrix.

iii. If matrix A has shape (m × n) and matrix B has shape (n × p), the resulting matrix will have shape (m × p).

iv. When applied to 1D tensors, it computes their dot product.

v. It is commonly used in machine learning operations such as neural network computations and linear algebra calculations

```python
# Element-wise multiplication (*)
a = torch.tensor([1, 2, 3])
b = torch.tensor([4, 5, 6])

print(f"Tensor A: {a}")
print(f"Tensor B: {b}")
print(f"Element-wise multiplication (A * B): {a * b}")

# Matrix multiplication (@)
mat_a = torch.tensor([[1, 2],
                      [3, 4]])
mat_b = torch.tensor([[5, 6],
                      [7, 8]])

print(f"\nMatrix A: {mat_a}")
print(f"Matrix B: {mat_b}")
print(f"Matrix multiplication (A @ B): {mat_a @ mat_b}")

# Example with 1D tensors (dot product)
vec_a = torch.tensor([1, 2, 3])
vec_b = torch.tensor([4, 5, 6])

print(f"\nVector A: {vec_a}")
print(f"Vector B: {vec_b}")
print(f"Dot product (Vector A @ Vector B): {vec_a @ vec_b}")
     
```

### Deeper Dive into reshape() vs view() Memory Behavior

The main difference between view() and reshape() is how they handle memory when changing the shape of a tensor.

#### tensor.view()

i. Always tries to create a new view of the original tensor without allocating additional memory.

ii. The new tensor shares the same memory as the original tensor.

iii. Any modification made to the view tensor is reflected in the original tensor, and vice versa.

iv. It only works when the original tensor is stored contiguously in memory.

v. If the tensor is not contiguous (for example, after a transpose operation), view() will produce an error unless .contiguous() is used first.

#### tensor.reshape()

i. Provides a more flexible way to change the shape of a tensor.

ii. It first attempts to return a view that shares memory with the original tensor.

iii. If a view cannot be created due to the tensor's memory layout, it automatically creates a copy in a new memory location.

iv. When a copy is created, changes made to the reshaped tensor do not affect the original tensor.

v. It can handle both contiguous and non-contiguous tensors, making it more convenient in many situations.

```python
import torch

t = torch.arange(10, 19, 1)  # values from 10 to 18

print(t.reshape(3, 3))
print(t.view(3, 3))
print(t)
```

### Moving to GPU in Machine Learning
Why GPUs are Preferred over CPUs-

#### CPU (Central Processing Unit)-

i.Designed for general-purpose computing

ii.Has few but powerful cores

iii.Best for sequential and complex tasks

iv.Works on tasks one after another
#### GPU (Graphics Processing Unit)-

i.Designed for parallel processing

ii. contains thousands of smaller cores

iii. Best for repetitive and large-scale computations

iv.Performs many operations simultaneously

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", device)

t = torch.tensor([1., 2., 3.])
t_gpu = t.to(device)
print(t_gpu.device)

# move back to CPU (needed before NumPy conversion)
t_cpu = t_gpu.cpu()
print(t_cpu.numpy())
```
